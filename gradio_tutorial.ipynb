## 2. 核心概念：Interface (gr.Interface)

gr.Interface 是 Gradio 的核心类。它将 Python 函数、输入组件和输出组件连接起来。

### 示例 1: 简单的文本处理 (Hello World)

这个函数接收一个字符串和一个布尔值，并返回一个问候语。

```python
def greet(name, is_uppercase):
    # 接收 name (str) 和 is_uppercase (bool)
    if is_uppercase:
        # 如果选中复选框，则使用大写
        greeting = f'HELLO, {name.upper()}!'
    else:
        # 否则使用常规格式
        greeting = f'您好，{name}！'
    return greeting

# inputs: Textbox (对应 name), Checkbox (对应 is_uppercase)
# outputs: Textbox (对应 greeting)
iface_text = gr.Interface(
    fn=greet, 
    inputs=[
        gr.Textbox(label='输入您的名字', placeholder='例如：小明'),
        gr.Checkbox(label='是否全部大写?')
    ], 
    outputs='text',
    title='Gradio 简单问候应用',
    description='演示基础的文本输入和布尔控制。'
)

# 运行应用：取消注释下一行
iface_text.launch()
```

## 3. 多输入与多输出 (Multiple Inputs and Outputs)

函数可以返回多个值，并映射到多个输出组件中。顺序必须严格对应。

### 示例 2: 简单计算器

```python
def calculator(num1, operation, num2):
    # num1 (float), operation (str), num2 (float)
    if operation == '加法 (+)':
        result = num1 + num2
    elif operation == '减法 (-)':
        result = num1 - num2
    elif operation == '乘法 (*)':
        result = num1 * num2
    elif operation == '除法 (/)':
        if num2 == 0:
            # 返回两个值：结果 (0) 和 错误消息
            return 0, '错误：除数不能为零！'
        result = num1 / num2
    else:
        return 0, '错误：无效的操作符'
    
    # 返回两个值：结果 (result) 和 状态消息 (str)
    return result, '操作成功完成！'

# inputs: Slider, Dropdown, Number
# outputs: Number (结果), Textbox (状态)
iface_calculator = gr.Interface(
    fn=calculator,
    inputs=[
        gr.Slider(0, 100, value=50, label='数字 1', step=1),
        gr.Dropdown(
            ['加法 (+)', '减法 (-)', '乘法 (*)', '除法 (/)'], 
            label='选择操作'
        ),
        gr.Number(label='数字 2', value=50)
    ],
    outputs=[
        gr.Number(label='结果'),
        gr.Textbox(label='状态信息', interactive=False)
    ],
    title='Gradio 简单计算器',
    description='演示了 Slider, Dropdown 和多输出的使用。'
)

# 运行应用：取消注释下一行
iface_calculator.launch()
```

## 4. 媒体组件：图像 (Image Component)

使用 gr.Image() 时，默认输入到 Python 函数的是一个 NumPy 数组，非常适合进行图像处理。

### 示例 3: 图像处理模拟

我们模拟一个实时（live=True）的图像滤镜功能。

```python
def image_filter(image_np, intensity):
    # image_np 是 NumPy 数组 (H, W, 3)
    # intensity 是浮点数 (0.0 到 1.0)
    
    # 将图像转换为浮点数并标准化 (0.0 到 1.0)
    img_float = image_np.astype(np.float32) / 255.0
    
    # 模拟滤镜效果：增加红色分量，减少蓝色分量 (强度受 intensity 控制)
    img_float[..., 0] = np.clip(img_float[..., 0] + intensity * 0.2, 0, 1)
    img_float[..., 2] = np.clip(img_float[..., 2] - intensity * 0.1, 0, 1)
    
    # 转换回 8 位整数 (0 到 255) 并返回
    processed_image = (img_float * 255).astype(np.uint8)
    
    return processed_image

# 创建 Interface
iface_image = gr.Interface(
    fn=image_filter,
    inputs=[
        gr.Image(
    label='上传一张图片',
    sources=['upload', 'webcam'],
    height=300,
    width=300
),
        gr.Slider(0, 1.0, value=0.5, label='滤镜强度', step=0.01)
    ],
    outputs=gr.Image(label='处理后的图片', type='numpy'),
    title='Gradio 图像滤镜模拟',
    description='开启 live=True，滑块调整时即时触发函数。',
    live=True # 开启实时模式
)

# 运行应用：取消注释下一行
iface_image.launch()
```

## 5. Blocks：高级布局与流程控制

使用 gr.Blocks() 可以实现更复杂的、自定义的布局和组件之间的事件连接。

### 示例 4: 使用 Blocks 创建分栏布局和手动事件连接

我们使用 gr.Row() 和 gr.Column() 创建一个两栏布局，并将 greet 和 image_filter 两个功能集成进来。

```python
with gr.Blocks(title='Gradio Blocks 综合示例') as demo_blocks:
    gr.Markdown('# Gradio Blocks 综合应用示例')
    gr.Markdown('--- Blocks 允许自定义布局和事件处理 ---')
    
    # 使用 Row 创建一个水平容器
    with gr.Row():
        
        # 左侧列 (占 1/3 宽度)
        with gr.Column(scale=1, variant='panel'): 
            gr.Markdown('## 模块 A: 问候应用')
            
            # 定义组件
            name_input = gr.Textbox(label='输入您的名字')
            uppercase_check = gr.Checkbox(label='是否全部大写?')
            greet_btn = gr.Button('生成问候语', variant='primary')
            greeting_output = gr.Textbox(label='问候语', interactive=False)
            
            # 绑定事件：当按钮被点击时，执行 greet 函数
            greet_btn.click(
                fn=greet, 
                inputs=[name_input, uppercase_check], 
                outputs=greeting_output
            )

        # 右侧列 (占 2/3 宽度)
        with gr.Column(scale=2, variant='panel'): 
            gr.Markdown('## 模块 B: 图像滤镜')
            
            # 定义组件
            img_in = gr.Image(label='上传图片', sources=['upload'],  height=300,
    width=300)
            intensity_slider = gr.Slider(0, 1.0, value=0.5, label='滤镜强度', step=0.01)
            img_out = gr.Image(label='处理后的图片', type='numpy')
            
            # 绑定事件：当滑块值变化时，执行 image_filter 函数
            intensity_slider.change(
                fn=image_filter,
                inputs=[img_in, intensity_slider],
                outputs=img_out,
            )
            # 绑定事件：当图片内容变化时 (例如用户上传新图)，也执行 image_filter
            img_in.change(
                fn=image_filter,
                inputs=[img_in, intensity_slider],
                outputs=img_out
            )

# 运行 Blocks 应用：取消注释下一行
demo_blocks.launch(share=True)
```
